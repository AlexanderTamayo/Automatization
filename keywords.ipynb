# Keywords

```python
import pandas as pd
import os

def keywords(input_file):
    input_file = input_file.strip('"').strip("'")
    input_file = os.path.normpath(input_file)
    
    df = pd.read_csv(input_file, skiprows=2)
    df = df.iloc[:-6, :]
    df = df[['Palabra clave', 'Impr.', 'Clics', 'Conversiones']]
    df['Conversiones'] = df['Conversiones'].astype(str).str.replace('.', '').str.replace(',', '.')
    df['Conversiones'] = df['Conversiones'].astype(float)
    df = df.groupby('Palabra clave')[['Impr.', 'Clics', 'Conversiones']].sum().reset_index()
    df_final = df.sort_values(by='Conversiones', ascending=False)
    return df_final.to_csv(input_file, index = False)

input_file = input("File Path: ")

keywords(input_file)
```
